```python
import PIL.Image as image
import matplotlib.pyplot as plt
import numpy as np
```

```python
# a) Read a grayscale image
gray_img = image.open(r"I:\My Drive\9th sem\DIP\Lab\images\hand_xray.jpg")
img = np.array(gray_img)
```

```python
# b) Apply image negative transformation Using formula: s = 255 - r
negative_img = 255 - img
```

```python
# c) Display both original and negative images
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Original Grayscale Image")
plt.imshow(gray_img, cmap='gray')
plt.axis('off')
plt.subplot(1,2,2)
plt.title("Negative Image (s = 255 - r)")
plt.imshow(negative_img, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```
