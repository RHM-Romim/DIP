```python
import PIL.Image as image
import matplotlib.pyplot as plt
import numpy as np
```

```python
# a) Read a grayscale image
gray_img = image.open(r'I:\My Drive\9th sem\DIP\Lab\images\greyscale_image.jpg')
gray_array = np.array(gray_img).astype('float32')
```

```python
# b) Increase the brightness of the image by multiplying each pixel value by S1 =1.5
s1 = 1.5
bright_img = gray_array * s1
bright_img = np.clip(bright_img, 0, 255).astype('uint8')
```

```python
# c) Decrease the brightness of the image by multiplying each pixel value by S2 = 0.3
s2 = 0.3
dark_img = gray_array * s2
dark_img = np.clip(dark_img, 0, 255).astype('uint8')
```

```python
# d) Compare the original grayscale image with the brightness-adjusted images
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(gray_array, cmap='gray', vmin=0, vmax=255)
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(bright_img, cmap='gray')
plt.title(f'Brightened (S={s1})')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(dark_img, cmap='gray')
plt.title(f'Darkened (S={s2})')
plt.axis('off')
plt.show()

```
