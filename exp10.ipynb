```python
import PIL.Image as image
import matplotlib.pyplot as plt
import numpy as np
```

```python
#a. Read a gray scale image
gray_img = image.open('sp_noise_img.png').convert('L')
img = np.array(gray_img)
```

```python

H,W= img.shape
plt.figure(figsize=(10,4))
plt.imshow(gray_img, cmap='gray')
plt.title("Origianl Image")
plt.axis('off')
```

```python
#b. Use replicate padding (edge padding) to handle boundary pixels
filter = np.array([
    [1/9,1/9,1/9],
    [1/9,1/9,1/9],
    [1/9,1/9,1/9]
])
```

```python

padded_img = np.pad(img,pad_width= 1, mode='edge')
```

```python
#c. Apply a 3×3 averaging filter to smooth the image
avg_opt=np.zeros_like(img)

for i in range(H):
    for j in range(W):
        region = padded_img[i:i+3,j:j+3]
        avg_opt[i,j]=np.sum(region * filter)
```

```python
#d. Apply a 3×3 median filter to smooth the image
mdn_opt=np.zeros_like(img)

for i in range(H):
    for j in range(W):
        region = padded_img[i:i+3,j:j+3]
        mdn_opt[i,j]=np.median(region)
```

```python
#e. Display the Original image, Averaging filter output, and Median filter output

plt.figure(figsize=(10,4))
plt.imshow(avg_opt, cmap='gray')
plt.title("After Average Filter")
plt.axis('off')

plt.figure(figsize=(10,4))
plt.imshow(mdn_opt, cmap='gray')
plt.title("After Median Filter")
plt.axis('off')
```
