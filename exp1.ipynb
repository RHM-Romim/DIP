```python
import PIL.Image as image
import matplotlib.pyplot as plt
import numpy as np

# a) Read an image
img_color = image.open(r'I:\My Drive\9th sem\DIP\Lab\images\japanese art.jpg')
# b) Display the image resolution in the format: Resolution = Width × Height pixels
img_width, img_height = img_color.size
print(f'Resolution = {img_width} × {img_height} pixels')
```

```python
# c) Print the image color mode
print(f'Color mode: {img_color.mode}')
```

```python
# d) Convert the image into grayscale
img_gray = img_color.convert('L')
print(f'{img_gray}')
```

```python
# e) Display both the original image and the grayscale image
# f) Label each displayed image with an appropriate title
plt.figure(figsize=(15, 8))
plt.subplot(1, 2, 1)
plt.imshow(img_color)
plt.title('Original Color Image')
plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Grayscale Image')
plt.show()
```

```python
# g) Save the grayscale image
img_gray.save(r'I:\My Drive\9th sem\DIP\Lab\images\img_gray.jpg')
print("Grayscale image saved in the folder!'")

```
