```python
import PIL.Image as image
import matplotlib.pyplot as plt
import numpy as np
```

```python
# a) Read a grayscale image
gray_img = image.open(r"I:\My Drive\9th sem\DIP\Lab\images\hand_xray.jpg")
img = np.array(gray_img)
```

```python
# b) Set a threshold value, k = 128
k = 128
```

```python
# c) Apply binary thresholding: if r > k → s = 255 else → s = 0
H, W = img.shape[:2]
img_threshold = np.zeros((H, W), dtype=np.uint8)
for i in range(H):
    for j in range(W):
        if img[i, j] > k:
            img_threshold[i, j] = 255
        else:
            img_threshold[i, j] = 0

```

```python
# d) Display the binary image
plt.figure(figsize=(12, 8))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(gray_img, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('Thresholded Image (k=128)')
plt.imshow(img_threshold, cmap='gray')
plt.show()
```
